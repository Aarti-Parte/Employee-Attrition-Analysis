```python
total_employees = len(df)
employees_left =(df["Attrition"]== "yes").sum()

attrition_rate = (employees_left / total_employees) * 100

print("Total Employees :", total_employees)
print("Employees_left:",employees_left)
print("Attrition Rate:",round(attrition_rate, 2),"%")
```

```python
print(df["Attrition"].unique())
print(df["Attrition"].value_counts())
```

```python

total_employees = df.shape[0]
employees_left = (df["Attrition"] == "Yes").sum()

attrition_rate = employees_left / total_employees * 100

print("Total Employees:", total_employees)
print("Employees Left:", employees_left)
print("Attrition Rate:", round(attrition_rate, 2), "%")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))

sns.countplot(data = df,x="Attrition")

plt.title("Employee Attrition Count")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

```python
department_attrition =df[df["Attrition"] == "Yes"].groupby("Department").size()

print(department_attrition)
```

```python
department_attrition = df[df["Attrition"] == "Yes"].groupby("Department").size()

plt.figure(figsize=(8,5))

department_attrition.plot(kind="bar")

plt.title("Employee  Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees Left")
plt.xticks(rotation = 0)

plt.show()
```

```python
total_by_department = df.groupby("Department").size()

left_by_department = df[df["Attrition"]== "Yes"].groupby("Department").size()

department_rate = (left_by_department / total_by_department * 100).round(2)

print("Department-wise Attrition rate (%):")
print(department_rate)
```

```python
plt.figure(figsize=(8,5))

department_rate.plot(kind = "bar")

plt.title("Department-wise Employee Attrition Rate")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation = 0)

plt.show()
```

```python
overtime_attrition = df.groupby("OverTime")["Attrition"].value_counts().unstack(fill_value =0)

print(overtime_attrition)
```

```python
overtime_rate =(
    df.groupby("OverTime")["Attrition"]
    .apply(lambda x:(x == "Yes").mean() * 100)
    .round(2)

)

print("Overtime-wise Attrition Rate (%):")
print(overtime_rate)
```

```python
plt.figure(figsize=(7,4))

overtime_rate.plot(kind="bar")


plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation = 0)

plt.show()
```
